```python
import numpy as np

X = np.array([[1, 1], [1, -1], [-1, 1], [-1, -1]])
Y = np.array([1, -1, -1, -1])
weights = np.zeros(2)
lr = 1

for i in range(len(X)):
    weights += lr * X[i] * Y[i]

print("Final weights:", weights)
```

```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]])
Y = np.array([0, 0, 0, 1])
weights = np.zeros(2)
bias = 0
lr = 1

for _ in range(10):
    for i in range(len(X)):
        y_pred = 1 if np.dot(X[i], weights) + bias >= 0 else 0
        error = Y[i] - y_pred
        weights += lr * error * X[i]
        bias += lr * error

print("Trained weights:", weights)
print("Bias:", bias)
```
